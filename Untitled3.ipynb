```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns  # Import seaborn for better visualizations

try:
    df = pd.read_csv("/Consumer_Complaints.csv")

    # 1. Data Cleaning (as before)
    df['Sub-product'] = df['Sub-product'].fillna('Unknown')
    df['Sub-issue'] = df['Sub-issue'].fillna('Not Applicable')
    df['Company public response'] = df['Company public response'].fillna('No response')
    df['Timely response?'] = df['Timely response?'].fillna('Unknown')
    df['Date submitted'] = pd.to_datetime(df['Date submitted'], errors='coerce')
    df['Date received'] = pd.to_datetime(df['Date received'], errors='coerce')
    df['Submitted via'] = df['Submitted via'].replace('Web Referral', 'Web')
    for col in ['Product', 'Sub-product', 'Issue', 'Sub-issue', 'Company public response', 'Company response to consumer']:
        df[col] = df[col].str.lower()
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()

    # 2. Feature Engineering
    df['Complaint Year'] = df['Date submitted'].dt.year
    df['Complaint Month'] = df['Date submitted'].dt.month

    # 3. Analysis and Visualization

    # a) Top Products and Issues
    plt.figure(figsize=(12, 6))
    df['Product'].value_counts().head(10).plot(kind='bar', color='skyblue')
    plt.title('Top 10 Products with Most Complaints')
    plt.xlabel('Product')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(12, 6))
    df['Issue'].value_counts().head(10).plot(kind='bar', color='lightcoral')
    plt.title('Top 10 Issues with Most Complaints')
    plt.xlabel('Issue')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # b) Complaint Volume Over Time
    plt.figure(figsize=(12, 6))
    df.groupby('Complaint Year')['Complaint ID'].count().plot(marker='o', color='mediumseagreen')
    plt.title('Complaint Volume Over Time')
    plt.xlabel('Year')
    plt.ylabel('Number of Complaints')
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # c) Companies with Highest Complaint Volume
    plt.figure(figsize=(12, 6))
    df['Company response to consumer'].value_counts().head(10).plot(kind='bar', color='gold')
    plt.title('Top 10 Company Response to Consumer')
    plt.xlabel('Company')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # d) Submission Method vs. Timely Response
    plt.figure(figsize=(8, 6))
    sns.countplot(x='Submitted via', hue='Timely response?', data=df, palette='Set2')
    plt.title('Submission Method vs. Timely Response')
    plt.xlabel('Submission Method')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # e) Sub-issues within Top Complaint Categories (example: 'incorrect information on your report')
    top_issue = 'incorrect information on your report'
    plt.figure(figsize=(12, 6))
    df[df['Issue'] == top_issue]['Sub-issue'].value_counts().head(10).plot(kind='bar', color='purple')
    plt.title(f'Top 10 Sub-issues for "{top_issue}"')
    plt.xlabel('Sub-issue')
    plt.ylabel('Number of Complaints')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

except FileNotFoundError:
    print("Error: File not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```
